```python
import numpy as np
import discretize
from discretize.utils import mesh_utils
import matplotlib.pyplot as plt
```

```python
# 1) Malla 

base_xy = 250.0
hx = mesh_utils.unpack_widths([(base_xy, 5, -1.3), (base_xy, 4), (base_xy, 5, 1.3)])
hy = hx.copy()

base_z = 150.0
hz = mesh_utils.unpack_widths([(base_z, 6, -1.3), (base_z, 10), (base_z, 6, 1.3)])

mesh = discretize.TensorMesh(
    [hx, hy, hz],
    x0=[-np.sum(hx) / 2, -np.sum(hy) / 2, -np.sum(hz[:10])]
)

# 2) Modelo 1D de conductividad
sigma = 1e-2 * np.ones(mesh.nC)
cc = mesh.cell_centers
x = cc[:, 0]
y = cc[:, 1]
z = cc[:, 2]

sigma[(z <= 0) & (z > -200)] = 1e-1
sigma[(z <= -200) & (z > -600)] = 5e-2
sigma[(z <= -600) & (z > -1200)] = 1e-2

sigma[z > 0] = 1e-8  # air

print("mesh =", mesh)
print("mesh.nC=", mesh.nC)
print("mesh.nEx=", mesh.nEx)
print("mesh.nEy=", mesh.nEy)
print("mesh.nEz=", mesh.nEz)
```

```python
# 3) Plot del modelo de conductividad 1D
mcollect_obj = mesh.plot_slice(np.log10(sigma), grid=True, normal="X")[0]
plt.colorbar(mcollect_obj, label="log10(sigma) [S/m]")
plt.title("Modelo de conductividad corte X")
plt.savefig(f"Model_Toy3_CorteX.png", dpi=300)
plt.show()
```

```python
mcollect_obj = mesh.plot_slice(np.log10(sigma), grid=True, normal="Y")[0]
plt.colorbar(mcollect_obj, label="log10(sigma) [S/m]")
plt.title("Modelo de conductividad corte Y")
plt.show()
```

```python
# 4) Guardar malla y modelo

mesh.save("mesh_Toy3_MT.json")
np.save("model_Toy3_MT.npy", sigma)
```
